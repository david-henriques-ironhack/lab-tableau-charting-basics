```python
import pandas as pd
```

```python
df = pd.read_csv("we_fn_use_c_marketing_customer_value_analysis.csv")
```

```python
df.dtypes
```

```python
# convert date
df["Effective To Date"] = pd.to_datetime(df["Effective To Date"])

# categorical columns
categorical_cols = [
    "State",
    "Response",
    "Coverage",
    "Education",
    "EmploymentStatus",
    "Gender",
    "Location Code",
    "Marital Status",
    "Policy Type",
    "Policy",
    "Renew Offer Type",
    "Sales Channel",
    "Vehicle Class",
    "Vehicle Size"
]

df[categorical_cols] = df[categorical_cols].astype("category")
```

```python
df["Customer"] = df["Customer"].astype("string")
```

```python
modified_df = df.copy()
modified_df.to_csv('modified_data.csv', index=False)
```
